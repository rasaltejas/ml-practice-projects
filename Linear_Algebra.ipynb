```python
#Matrix
# Rows = students
# Columns = subjects
import numpy as np
students = np.array([
    [80, 50, 80],
    [80, 70, 90],
    [50, 60, 30]]
)
print(students)
print(students.shape)
print(students.ndim)  #used for check dimensions
```

```python
import numpy as np

# Calculate the total marks for each student (sum across rows)
total_marks_per_student = np.sum(students, axis=1)
print("Total marks per student:")
print(total_marks_per_student)
```
